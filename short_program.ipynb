```python
ls1=["name","age","subject"]
ls2=[]
for i in range(len(ls1)):
    name=input("enter your name")
    ls2.append(name)
    age=int(input("enter your age"))
    ls2.append(age)
    sub=input("enter your subject")
    ls2.append(sub)
    di=dict(zip(ls1,ls2))
print(di)
```
