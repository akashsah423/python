## found a maximum number

```python
ls=[2,3,4,5,6,7,1,9,234]
maximum=0
for i in range(0,9,1):
    if ls[i]>maximum:
        maximum=ls[i]
print(maximum)
        
```

## found a minimum number

```python
ls=[2,3,4,5,6,7,1,9,234]
minimum=1
for i in range(0,9,1):
    if ls[i]<minimum:
        minimum=ls[i]
print(minimum)
                
```

## count number 

```python
num=124569876
count=0
while num>0:
    num=num//10
    count+=1
print(count)
```

## addition of all digit in a number

```python
num=124569876
count=0
add=0
while num>0:
    temp=num%10
    num=num//10
    count+=1
    add=add+temp
print(count)
print(add)
```

## nested loops

```python
for i in range(1,11):
    print("student",i)
    for marks in range(1,5):
        print(marks,"marks",end=" , ")
    print()
```

```python
for i in range(1,11):
    print("student",i)
    
    for marks in range(1,i+1):
        print("marks",marks,end="   ")
    print()     
```

```python
for i in range(1,11):
    print("student",i)
    
    for marks in range(i+1,11):
        print("marks",marks)
    
```
